```python
classes = df["Class"].unique().tolist()
ratios = []
for course in classes:
    tempdf = df[df["Class"] == course]
    good = tempdf[tempdf["Online Course"] == 1]["Online Course"].count()
    bad = tempdf[tempdf["Online Course"] == -1]["Online Course"].count()
    case = tempdf[tempdf["Online Course"] == 0]["Online Course"].count()
    print("class:", course)
    print("good count:", good)
    print("bad count:", bad)
    print("case-by-case count:", case)
    print(f"good ratio: {good/(tempdf["Online Course"].count()):.3f}")
    print(f"bad ratio: {bad/(tempdf["Online Course"].count()):.3f}")
    print(f"case-by-case ratio: {case/(tempdf["Online Course"].count()):.3f}")
    ratios.append(case/(tempdf["Online Course"].count()))
    print("")
ratios
h, p = kruskal(0.1544, 0.251, 0.271, 0.1485, 0.2279, 0.260)
print(p)
```

```python
bchm = df[(df["Course"] == "Biochemistry")]
bchm1 = bchm.groupby(["State", "Online Course"]).size().reset_index(name = "count")
for index, row in bchm1.iterrows():
    if row["Online Course"] == 0.0:
        bchm1.loc[index, "Online Course"] = "Case-by-case"
    if row["Online Course"] == 1.0:
        bchm1.loc[index, "Online Course"] = "Approved"
    if row["Online Course"] == -1.0:
        bchm1.loc[index, "Online Course"] = "Not approved"

pivot = bchm1.pivot(index="State", columns="Online Course", values="count").fillna(0)
pivot_masked = pivot.mask(pivot == 0)
fig = px.imshow(
    pivot_masked,
    aspect="auto",
    labels={"color": "Count"},
    title="Number of Medical Schools' Online Biochemistry Course Approval by State",
    color_continuous_scale="Plotly3"
)
 
fig.update_layout(width=800, height=900)
```
